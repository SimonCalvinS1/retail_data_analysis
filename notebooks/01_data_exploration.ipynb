## Data Cleaning

```python
df = pd.read_excel('../data/retail_dataset.xlsx')

# Fix unclean data: (State -> [Region, City])
state_geography_map = {
    'Andhra Pradesh': ['South', 'Hyderabad'],
    'Bihar': ['East', 'Patna'],
    'Chhattisgarh': ['Central', 'Bhopal'], 
    'Delhi': ['North', 'Delhi'],
    'Goa': ['West', 'Panaji'],
    'Gujarat': ['West', 'Ahmedabad'],
    'Haryana': ['North', 'Chandigarh'],
    'Himachal Pradesh': ['North', 'Shimla'],
    'Jharkhand': ['East', 'Ranchi'],
    'Karnataka': ['South', 'Bangalore'],
    'Kerala': ['South', 'Kochi'],
    'Madhya Pradesh': ['Central', 'Bhopal'],
    'Maharashtra': ['West', 'Mumbai'],
    'Odisha': ['East', 'Bhubaneswar'],
    'Punjab': ['North', 'Chandigarh'],
    'Rajasthan': ['North', 'Jaipur'],
    'Tamil Nadu': ['South', 'Chennai'],
    'Telangana': ['South', 'Hyderabad'],
    'Uttar Pradesh': ['North', 'Lucknow'],
    'West Bengal': ['East', 'Kolkata']
}

def fix_data(row):
    state = row['State']
    if state in state_geography_map:
        row['region'] = state_geography_map[state][0]
        row['City'] = state_geography_map[state][1]
    return row

df = df.apply(fix_data, axis=1)

df['invoice_date'] = pd.to_datetime(df['invoice_date']).dt.date

df.columns = df.columns.str.strip()

df.to_excel('../data/retail_dataset_cleaned.xlsx', index=False)
print("Data Cleaned and Saved to 'retail_dataset_cleaned.xlsx'")
```

```python
df = pd.read_excel('../data/retail_dataset_cleaned.xlsx')

df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
print("Standardized Columns:\n", df.columns.tolist())

df['invoice_date'] = pd.to_datetime(df['invoice_date']).dt.date
df.to_excel('../data/retail_dataset_cleaned.xlsx', index=False)
```

```python
df['sales_method'] = df['sales_method'].str.rstrip(';') # remove trailing semicolons
df.to_excel('../data/retail_dataset_cleaned.xlsx', index=False)
print("Trailing semicolons removed")
```
