## Module imports with loading the dataset:

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel('../data/retail_dataset_cleaned.xlsx')
print("Modules imported successfully and data loaded")
```

## Top Products based on Revenue:

```python
top_products_revenue = df.groupby('product')['total_sales'].sum().sort_values(ascending=False).head(10)
print("\nTop 10 Products by Revenue:")
print(top_products_revenue)
```

## Top Products by Quantity Sold:

```python
top_products_quantity = df.groupby('product')['units_sold'].sum().sort_values(ascending=False).head(10)
print("\nTop 10 Products by Quantity Sold:")
print(top_products_quantity)
```

## Category-wise Performance:

```python
category_sales = df.groupby('sales_method')['total_sales'].sum().sort_values(ascending=False)
print("\nSales by Category:")
print(category_sales)
```

## Visualizations of Top Sales and Category based Sales:

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

top_products_revenue.plot(kind='barh', ax=axes[0], color='steelblue')
axes[0].set_title('Top 10 Products by Revenue', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Revenue (Rs.)')

category_sales.plot(kind='pie', ax=axes[1], autopct='%1.1f%%', startangle=90)
axes[1].set_title('Revenue Distribution by Category', fontsize=14, fontweight='bold')
# axes[1].set_ylabel('')

plt.tight_layout()
plt.show()
```

## Revenue vs Quantity (Pandas Split-Apply-Combine operation pipeline)

```python
product_performance = df.groupby('product').agg({
    'total_sales': 'sum',
    'units_sold': 'sum',
    'retailer_id': 'count'
}).rename(columns={'retailer_id': 'num_retailers'})

product_performance['Avg_Price'] = product_performance['total_sales'] / product_performance['units_sold']

print("\nProduct Performance Matrix:")
print(product_performance.sort_values('total_sales', ascending=False).head(15))
```
