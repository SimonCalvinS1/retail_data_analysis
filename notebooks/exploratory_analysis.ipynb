## Module imports:

```python
import pandas as pd
```

## Data Cleaning

```python
df = pd.read_excel('../data/retail_dataset.xlsx')

city_to_state = {
    'Bangalore': 'Karnataka',
    'Jaipur': 'Rajasthan',
    'Chandigarh': 'Chandigarh',
    'Lucknow': 'Uttar Pradesh',
    'Ahmedabad': 'Gujarat',
    'Hyderabad': 'Telangana',
    'Mumbai': 'Maharashtra',
    'Pune': 'Maharashtra',
    'Chennai': 'Tamil Nadu',
    'Kolkata': 'West Bengal',
    'Bhopal': 'Madhya Pradesh',
    'Delhi': 'Delhi',
    'Patna': 'Bihar'
}

state_to_region = {
    'Karnataka': 'South', 
    'Telangana': 'South', 
    'Tamil Nadu': 'South',
    'Maharashtra': 'West', 
    'Gujarat': 'West',
    'Rajasthan': 'North', 
    'Delhi': 'North', 
    'Chandigarh': 'North', 
    'Uttar Pradesh': 'North',
    'Madhya Pradesh': 'Central',
    'West Bengal': 'East', 
    'Bihar': 'East'
}

def fix_geography(row):
    city = row['City']
    if city in city_to_state:
        new_state = city_to_state[city]
        new_region = state_to_region.get(new_state, "Other")
        return pd.Series([new_state, new_region])
    elif city == "Unknown City" or pd.isnull(city):
        return pd.Series(['Unknown State', 'Unknown Region'])
    
    return pd.Series([row['State'], row['region']])

df[['State', 'region']] = df.apply(fix_geography, axis=1)

df['invoice_date'] = pd.to_datetime(df['invoice_date'])
df['invoice_date'] = df['invoice_date'].dt.date # Convert to date only, removing time component

df.to_excel('../data/retail_dataset_cleaned.xlsx', index=False)
print("Data Cleaned and Saved to 'retail_dataset_cleaned.xlsx'")
```
